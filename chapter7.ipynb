```python
data = pd.DataFrame({"food": ["bacon", "pulled_pork", "bacon", "pastrami", "corned beef", "bacon", "pastrami", "honey ham", "nova lox"], 
                     "ounces": [4, 3, 12, 6, 7.5, 8, 3, 5, 6]})
data
```

```python
meat_to_animal = {
  "bacon": "pig",
  "pulled pork": "pig",
  "pastrami": "cow",
  "corned beef": "cow",
  "honey ham": "pig",
  "nova lox": "salmon"
}
```

```python
data["animal"] = data["food"].map(meat_to_animal)
data
```
