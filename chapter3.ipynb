```python
words = ["apple", "bat", "bar", "atom", "book"]

by_letter = {}

for word in words:
    letter = word[0]
    if letter not in by_letter:
        by_letter[letter] = [word]
    else:
        by_letter[letter].append(word)

by_letter
```

```python
by_letter = {}

for word in words:
    letter = word[0]
    by_letter.setdefault(letter, []).append(word)

by_letter
```

```python
def _make_gen(x):
    for i in range(x):
        yield 2 * i

gen = _make_gen(100)

for i in gen:
    print(i)
```
